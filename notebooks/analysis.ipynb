```python
import pandas as pd

df = pd.read_csv("../results/benchmark_results.csv")
df.head()
```

```python
df.groupby("model")[["wer", "inference_time", "rtf"]].mean()
```

```python
# Worst-performing samples per model — useful for error analysis
df.sort_values("wer", ascending=False).groupby("model").head(3)
```

```python
import matplotlib.pyplot as plt

df.boxplot(column="wer", by="model")
plt.title("WER distribution by model")
plt.suptitle("")
plt.ylabel("WER")
plt.show()
```
